```python
monthly_category_sales = df.pivot_table(index='Month', columns='category', values='Sales', aggfunc='sum')

month_order = ['January','February','March','April','May','June','July','August','September','October','November','December']
monthly_category_sales = monthly_category_sales.reindex(month_order)
```

```python
months = monthly_category_sales.index
categories = monthly_category_sales.columns
num_months = len(months)
num_categories = len(categories)

bar_width = 0.8 / num_categories  
positions = range(num_months)

plt.figure(figsize=(12,6))

for i, category in enumerate(categories):
    plt.bar([p + i*bar_width for p in positions], monthly_category_sales[category], width=bar_width, label=category)

# Labels and title
plt.xticks([p + bar_width*(num_categories-1)/2 for p in positions], months)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales by Category')
plt.legend()
plt.show()
```

The analysis shows that a few products generate the majority of sales, with Electronics and Fashion dominating. Credit Card is the preferred payment method, while Easypaisa and JazzCash are less used. Karachi and Lahore are the top-performing cities. Sales peak during December, indicating seasonal demand. These insights can help optimize inventory, marketing campaigns, and payment options to improve revenue and customer experience

This bar chart shows monthly sales across three categories: Accessories, Electronics, and Fashion.

# Conclusion based on the graph:

Accessories have strong sales in the first four months (January to April) and March is especially high. However, sales decline significantly mid-year and remain relatively low from June onwards.

Electronics sales are more stable throughout the year, with peaks around March, May, and October. This category generally maintains moderate to high sales every month.

Fashion shows a more fluctuating trend with notable peaks in February, May, and August, and generally moderate sales in other months.

# Overall:

Accessories dominate early in the year but slow down mid-year.

Electronics sustain consistent sales with some mid-year and fall spikes.

Fashion shows varied demand, with intermittent spikes but less consistency than Electronics.

These insights suggest targeted marketing or inventory strategies could focus on boosting Accessories sales mid-to-late year, while maintaining steady stock of Electronics year-round, and leveraging Fashion's peaks with promotions during February, May, and August.
